## Load Libraries

I am using pandas for dataframe work, numpy for missing values, and regular expressions for cleaning column names.

```python
import pandas as pd
import numpy as np
import re

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## Q1(a) Store the Data as a Pandas DataFrame and Examine Datatypes

### In this section, I load the Fitbit activity dataset from the course GitHub repository into a pandas dataframe. After loading the data, I inspect the first few rows and check the datatype of each column.

### This is important because some columns may load as text even though they represent dates, numbers, or time-based values.

```python
url = "https://raw.githubusercontent.com/nurfnick/Data_Viz/refs/heads/main/Data_Sets/Activity_Dataset_V1.csv"

activity_raw = pd.read_csv(url)

print("Rows and columns:", activity_raw.shape)
display(activity_raw.head())

dtype_table = pd.DataFrame({
    "column": activity_raw.columns,
    "dtype": activity_raw.dtypes.astype(str).values,
    "non_null_count": activity_raw.notna().sum().values,
    "missing_count": activity_raw.isna().sum().values
})

display(dtype_table)
```

#### The dataset was successfully stored as a pandas dataframe. The datatype table shows that some columns are already numeric, while others need conversion. For example, `activity day` should be converted into datetime format because it represents dates. The `avg pace` column may need cleaning because pace values can be stored as text. Numeric columns such as `total steps`, `calories`, `aerobic`, and `max cadence` should be numeric for calculations.

## Q1(b) Remove the Column that Repeats the Indexes

### The dataset contains an unnamed column that repeats the dataframe index. Since pandas already provides an index, this column is unnecessary and should be removed.

```python
activity = activity_raw.copy()

unnamed_cols = activity.columns[activity.columns.str.contains(r"^Unnamed", case=False, regex=True)]

print("Columns removed:", list(unnamed_cols))

activity = activity.drop(columns=unnamed_cols)

display(activity.head())
```

#### The unnamed index column was removed from the dataframe. This makes the dataset cleaner because the repeated index column does not provide useful information for analysis.

## Q1(c) Clean Column Names to Remove the Unit Declaration ` (%)`

### In this section, I clean the column names using regular expressions. The exam specifically asks to remove the unit declaration ` (%)` using regex. I also remove extra spaces and convert column names to lowercase to make them easier to use in code.

```python
print("Before cleaning:")
print(list(activity.columns))

activity.columns = (
    activity.columns
    .str.replace(r"\s*\(%\)", "", regex=True)   # removes (%)
    .str.replace(r"\(%\)", "", regex=True)      # removes (%) when no space exists
    .str.strip()
    .str.lower()
)

print("\nAfter cleaning:")
print(list(activity.columns))
```

#### The column names were cleaned using a regular expression. The ` (%)` unit declaration was removed, trailing spaces were removed, and the column names were standardized to lowercase. This makes the dataframe easier to work with in later sections.

## Q1(d) Convert `activity day` Column into Datetime Format

### The `activity day` column represents dates, so it should be converted into pandas datetime format. This allows date-based operations, such as finding the day of the week.

```python
activity["activity_day"] = pd.to_datetime(activity["activity_day"], errors="coerce")

print("Datatype after conversion:", activity["activity_day"].dtype)
display(activity[["activity_day"]].head())
```

## Q1(e) Impute `total_steps` and Convert to Appropriate Datatype

### The `total_steps` column contains step-count data, so it should be numeric. If there are missing values, I replace them with the median number of steps. I use the median because step counts can have very high or very low outliers, and the median is less affected by extreme values than the mean.

### After filling the missing values, I convert the column to integer because steps are whole-number counts.

```python
activity["total_steps"] = pd.to_numeric(activity["total_steps"], errors="coerce")

missing_before = activity["total_steps"].isna().sum()
median_steps = activity["total_steps"].median()

activity["total_steps"] = activity["total_steps"].fillna(median_steps).round().astype(int)

missing_after = activity["total_steps"].isna().sum()

print("Missing total_steps before imputation:", missing_before)
print("Median used for imputation:", median_steps)
print("Missing total_steps after imputation:", missing_after)
print("Datatype after conversion:", activity["total_steps"].dtype)

display(activity[["total_steps"]].head())
```

#### The missing values in `total_steps` were replaced with the median step count. This keeps all rows in the dataset while filling missing values with a reasonable number. The column was converted to integer format because step counts should be whole numbers.

## Q1(f) Convert Non-Empty `avg_pace` into a Float

### The `avg_pace` column contains pace information. Some values may be written in a minutes:seconds format. For example, `3:30` means 3 minutes and 30 seconds, which is equal to 3.5 minutes.

### This conversion changes the pace values into decimal minutes so the column can be used as numeric data.

```python
def pace_to_minutes(value):
    """
    Convert pace values like '3:30' into decimal minutes.
    Example: 3:30 becomes 3.5 minutes.
    """
    if pd.isna(value):
        return np.nan
    
    text = str(value).strip()
    
    if text == "" or text.lower() in {"nan", "none"}:
        return np.nan
    
    if ":" in text:
        parts = text.split(":")
        
        try:
            parts = [float(part) for part in parts]
        except ValueError:
            return np.nan
        
        if len(parts) == 2:
            minutes, seconds = parts
            return minutes + seconds / 60
        
        elif len(parts) == 3:
            hours, minutes, seconds = parts
            return hours * 60 + minutes + seconds / 60
    
    return pd.to_numeric(text, errors="coerce")


activity["avg_pace"] = activity["avg_pace"].apply(pace_to_minutes)

print("Datatype after conversion:", activity["avg_pace"].dtype)
display(activity[["avg_pace"]].head(10))
```

#### The non-empty `avg_pace` values were converted into float values representing decimal minutes. Values written in minutes and seconds are now easier to analyze numerically. Missing or empty pace values remain as missing values.

## Q1(g) Group Data by `workout_type` and Find Mean, Median, Count, and Standard Deviation of `calories`

### In this section, I group the data by `workout_type` and calculate summary statistics for `calories`. The mean and median show typical calorie burn, the count shows how many activities are in each workout type, and the standard deviation shows how spread out the calorie values are.

```python
activity["calories"] = pd.to_numeric(activity["calories"], errors="coerce")

calorie_summary = (
    activity
    .groupby("workout_type")["calories"]
    .agg(["mean", "median", "count", "std"])
    .reset_index()
    .sort_values("mean", ascending=False)
)

display(calorie_summary)
```

#### The summary table shows the mean, median, count, and standard deviation of `calories` for each `workout_type`. This helps compare calorie burn across different types of workouts.

## Q1(h) Create an Indicator Column for 30% or More `aerobic` Activity

### The `aerobic` column represents the percentage of aerobic activity. I create a new indicator column that shows whether an activity achieved 30% or more aerobic activity.

### If the value is 30 or greater, the indicator is `True`. If the value is below 30, the indicator is `False`.

```python
activity["aerobic"] = pd.to_numeric(activity["aerobic"], errors="coerce")

activity["aerobic_30_percent_indicator"] = activity["aerobic"] >= 30

display(activity[[
    "activity_day",
    "workout_type",
    "aerobic",
    "aerobic_30_percent_indicator"
]].head(10))

indicator_counts = activity["aerobic_30_percent_indicator"].value_counts(dropna=False)

display(indicator_counts)
```

#### The indicator column was created successfully. It identifies which activities achieved at least 30% aerobic activity. This makes it easier to separate workouts that met the aerobic threshold from those that did not.

## Q1(i) Find the Day of the Week and `workout_type` with the Maximum `max_cadence`

### In this section, I use the datetime version of `activity_day` to create a new column called `day_of_week`. Then I find the row with the highest `max_cadence` value and display the activity date, day of the week, workout type, and max cadence.

```python
activity["max_cadence"] = pd.to_numeric(activity["max_cadence"], errors="coerce")

activity["day_of_week"] = activity["activity_day"].dt.day_name()

max_cadence_index = activity["max_cadence"].idxmax()

max_cadence_result = activity.loc[
    max_cadence_index,
    ["activity_day", "day_of_week", "workout_type", "max_cadence"]
]

print("The row with the maximum max_cadence is:")
display(max_cadence_result.to_frame(name="value"))
```

#### The output shows the record with the maximum `max_cadence`. It includes the date, day of the week, workout type, and cadence value. This directly answers which day and workout type had the highest maximum cadence.
